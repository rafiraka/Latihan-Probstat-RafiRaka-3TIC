```python
print("=== KASUS 1: ANALISIS KEBERHASILAN REQUEST BERDASARKAN JARINGAN ===")
total_wifi = 200
berhasil_wifi = 180
total_seluler = 200
berhasil_seluler = 160

# Menghitung probabilitas
prob_berhasil_wifi = berhasil_wifi / total_wifi
prob_berhasil_seluler = berhasil_seluler / total_seluler

print("Langkah 1: Identifikasi Data Dari Tabel")
print(f"  - Total Request Wi-Fi      = {total_wifi}")
print(f"  - Request Berhasil (Wi-Fi) = {berhasil_wifi}")
print(f"  - Total Request Seluler    = {total_seluler}")
print(f"  - Request Berhasil (Sel)   = {berhasil_seluler}\n")

print("Langkah 2: Menhitung Probabilitas Bersyarat P(Berhasil | Jaringan)")
print(f"  a. P(Berhasil | Wi-Fi)   = Berhasil Wi-Fi / Total Wi-Fi")
print(f"                           = {berhasil_wifi} / {total_wifi}")
print(f"                           = {prob_berhasil_wifi:.2f} ({prob_berhasil_wifi*100:.0f}%)\n")

print(f"  b. P(Berhasil | Seluler) = Berhasil Seluler / Total Seluler")
print(f"                           = {berhasil_seluler} / {total_seluler}")
print(f"                           = {prob_berhasil_seluler:.2f} ({prob_berhasil_seluler*100:.0f}%)\n")

print("Langkah 3: Kesimpulan")
print("  Jaringan Wi-Fi memiliki tingkat keberhasilan lebih tinggi (90%) dibanding Seluler (80%).")
print("="*65 + "\n")
```

```python
print("=== KASUS 2: PENGUJIAN MODUL PERANGKAT LUNAK (TEOREMA BAYES) ===")
p_bug = 0.10         # P(B)
p_no_bug = 0.90      # P(B^c)
p_pos_bug = 0.80     # P(P|B)
p_pos_no_bug = 0.15  # P(P|B^c)
# Hitung komponen
pos_dan_bug = p_pos_bug * p_bug
pos_dan_nobug = p_pos_no_bug * p_no_bug
p_pos_total = pos_dan_bug + pos_dan_nobug
p_bug_given_pos = pos_dan_bug / p_pos_total

print("Langkah 1: Identifikasi Probabilitas Awal")
print(f"  - P(B)   [Modul Ber-bug]       = {p_bug}")
print(f"  - P(B^c) [Modul Tanpa Bug]     = {p_no_bug}")
print(f"  - P(P|B) [Sensitivitas Alat]   = {p_pos_bug}")
print(f"  - P(P|B^c) [False Positive]    = {p_pos_no_bug}\n")

print("Langkah 2: Hitung Probabilitas Total Terdeteksi Positif P(P)")
print(f"  - Positif Asli  = P(P|B) * P(B)     = {p_pos_bug} * {p_bug} = {pos_dan_bug:.4f}")
print(f"  - Positif Palsu = P(P|B^c) * P(B^c) = {p_pos_no_bug} * {p_no_bug} = {pos_dan_nobug:.4f}")
print(f"  - Total P(P)    = {pos_dan_bug:.4f} + {pos_dan_nobug:.4f} = {p_pos_total:.4f}\n")

print("Langkah 3: Hitung Teorema Bayes P(B|P)")
print(f"  - P(B|P) = [P(P|B) * P(B)] / P(P)")
print(f"           = {pos_dan_bug:.4f} / {p_pos_total:.4f}")
print(f"           = {p_bug_given_pos:.4f} ({p_bug_given_pos*100:.2f}%)\n")

print("Langkah 4: Kesimpulan")
print("  Probabilitas modul benar-benar memiliki bug kritis setelah dinyatakan positif adalah 37.21%.")
print("="*65 + "\n")
```

```python
print("=== KASUS 3: ANALISIS INSIDEN PADA APLIKASI WEB ===")
total_request = 1000
n_timeout = 80
n_http500 = 50
n_keduanya = 20

n_gabungan = n_timeout + n_http500 - n_keduanya
prob_timeout_atau_500 = n_gabungan / total_request

print("Langkah 1: Identifikasi Data Insiden")
print(f"  - Total Request             = {total_request}")
print(f"  - Timeout          [n(A)]   = {n_timeout}")
print(f"  - HTTP 500         [n(B)]   = {n_http500}")
print(f"  - Keduanya         [n(A∩B)] = {n_keduanya}\n")

print("Langkah 2: Terapkan Prinsip Inklusi-Eksklusi")
print(f"  - n(A U B) = n(A) + n(B) - n(A ∩ B)")
print(f"             = {n_timeout} + {n_http500} - {n_keduanya}")
print(f"             = {n_gabungan} request bermasalah\n")

print("Langkah 3: Hitung Probabilitas Gabungan P(A U B)")
print(f"  - P(A U B) = n(A U B) / Total Request")
print(f"             = {n_gabungan} / {total_request}")
print(f"             = {prob_timeout_atau_500:.2f} ({prob_timeout_atau_500*100:.0f}%)\n")

print("Langkah 4: Kesimpulan")
print("  Probabilitas sebuah request mengalami timeout atau HTTP 500 adalah 11%.")
print("="*65)
```
